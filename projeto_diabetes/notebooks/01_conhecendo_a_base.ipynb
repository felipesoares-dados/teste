O diabetes é uma doença crônica grave na qual os indivíduos perdem a capacidade de regular efetivamente os níveis de glicose no sangue e pode levar a uma redução na qualidade de vida e na expectativa de vida.

O Sistema de Vigilância de Fatores de Risco Comportamentais (BRFSS) é uma pesquisa telefônica relacionada à saúde que é coletada anualmente pelo CDC (Centro de Controle e Prevenção de Doenças dos Estados Unidos). A cada ano, a pesquisa coleta respostas de milhares de americanos sobre comportamentos de risco relacionados à saúde, condições crônicas de saúde e o uso de serviços preventivos. Para este projeto, foi utilizado conjunto de dados disponível no Kaggle para o ano de 2015. 

https://www.kaggle.com/datasets/alexteboul/diabetes-health-indicators-dataset

Dicionário de variáveis:

- `Diabetes_binary`: 0 = sem diabetes, 1 = com diabetes
- `HighBP`: 0 = sem pressão alta, 1 = com pressão alta
- `HighChol`: 0 = sem colesterol alto, 1 = com colesterol alto
- `CholCheck`: 0 = não fez exame de colesterol na vida, 1 = fez exame de colesterol alguma vez
- `BMI`: Índice de Massa Corporal (IMC)
- `Smoker`: 0 = não fumante, 1 = fumante
- `Stroke`: 0 = sem histórico de AVC, 1 = com histórico de AVC
- `HeartDiseaseorAttack`: 0 = sem histórico de doença cardíaca ou ataque cardíaco, 1 = com histórico de doença cardíaca ou ataque cardíaco
- `PhysActivity`: 0 = não pratica atividade física, 1 = pratica atividade física
- `Fruits`: 0 = não consome frutas, 1 = consome frutas
- `Veggies`: 0 = não consome vegetais, 1 = consome vegetais
- `HvyAlcoholConsump`: 0 = não consome álcool em altas quantidades, 1 = consome álcool em altas quantidades
- `AnyHealthcare`: 0 = não tem plano de saúde, 1 = tem plano de saúde
- `NoDocbcCost`: 0 = não foi ao médico por questões financeiras, 1 = foi ao médico por questões financeiras (últimos 12 meses)
- `GenHlth`: Saúde geral (1 a 5) - 1 = Excelente, 2 = Muito boa, 3 = Boa, 4 = Aceitável, 5 = Ruim
- `MentHlth`: Nos últimos 30 dias, quantos dias a saúde mental não foi boa (0 a 30)
- `PhysHlth`: Nos últimos 30 dias, quantos dias a saúde física não foi boa (0 a 30)
- `DiffWalk`: 0 = não tem dificuldade para caminhar, 1 = tem dificuldade para caminhar
- `Sex`: 0 = feminino, 1 = masculino
- `Age`: Idade em faixas 1 = 18-24; 2 = 25-29; 3 = 30-34; 4 = 35-39; 5 = 40-44; 6 = 45-49; 7 = 50-54; 8 = 55-59; 9 = 60-64; 10 = 65-69; 11 = 70-74; 12 = 75-79; 13 = 80+
- `Education`: Níveis de ensino 1 = nunca frequentou a escola; 2 = escola primária; 3 = escola secundária incompleta ; 4 = escola secundária; 5 = faculdade incompleta ou curso técnico; 6 = completou faculdade ou diplomas superiores
- `Income`: Renda anual em faixas 1 = < 10.000; 2 = 10.000-14.999; 3 = 15.000-19.999; 4 = 20.000-24.999; 5 = 25.000-34.999; 6 = 35.000-49.999; 7 = 50.000-74.999; 8 = 75.000+

```python
import sys
from pathlib import Path

import pandas as pd

# Os notebooks ficam em notebooks/. Adicionamos a raiz do projeto ao caminho de
# importação para que o pacote `src` (config, estatistica, graficos) seja encontrado.
sys.path.append(str(Path.cwd().parent))

from src.config import (
    DADOS_ORIGINAIS,
    DADOS_TRATADOS,
)
```

## Carregando os dados

Leitura do arquivo compactado (`.zip`) cujo caminho está definido em `src/config.py`.

```python
# O pandas lê o CSV direto de dentro do .zip (compression='zip')
df_diabetes = pd.read_csv(DADOS_ORIGINAIS, compression='zip')
```

Primeiras linhas da base com `head()`. Por padrão o pandas oculta colunas quando há muitas; com `option_context` exibimos todas elas apenas nesta visualização, sem alterar a configuração global.

```python
with pd.option_context("display.max_columns", None):
    display(df_diabetes.head())
```

O `info()` mostra a estrutura da base: 70.692 registros e 22 colunas. As colunas numéricas, inclusive as binárias, vêm como `float64`.

```python
df_diabetes.info()
```

## Renomeando as colunas

Traduzimos os nomes originais (em inglês) para português, na mesma ordem das colunas da base, para facilitar a leitura das análises.

```python
# Nomes em português, na mesma ordem das colunas originais (ver referencias/01_dicionario_de_dados.md)
df_diabetes.columns = [
    "Diabetes",
    "PressaoAlta",
    "ColesterolAlto",
    "ColesterolExame",
    "IMC",
    "Fumante",
    "AVC",
    "ProblemaCardiaco",
    "AtividadeFisica",
    "ComeFrutas",
    "ComeLegumes",
    "ConsumoBebidaAlcoolica",
    "PlanoSaude",
    "SemDinheiroConsultas",
    "SaudeGeral",
    "DiasProblemasMentais",
    "DiasProblemasFisicos",
    "DificuldadeAndar",
    "Sexo",
    "FaixaIdade",
    "Ensino",
    "FaixaRenda",
]
```

O `nunique()` conta os valores distintos de cada coluna. Com isso identificamos o tipo de cada variável: binárias (2 valores), categóricas ordinais (de 5 a 13 valores) e numéricas (`IMC` com 80 valores e os dias de problemas com 31).

```python
df_diabetes.nunique()
```

Seleciona as colunas com exatamente 2 valores distintos, ou seja, as binárias. São 15, contando `Diabetes` e `Sexo`.

```python
# Colunas com exatamente 2 valores únicos são as binárias (inclui Diabetes e Sexo)
colunas_binarias = df_diabetes.nunique()[df_diabetes.nunique() == 2].index.to_list()

colunas_binarias
```

## Tratamento dos tipos de dados

Criamos uma cópia (`df_diabetes_tratado`) para preservar o DataFrame original e aplicar as transformações somente nela.

```python
df_diabetes_tratado = df_diabetes.copy()

df_diabetes_tratado.info()
```

Converte cada coluna binária para categórica, trocando 0/1 por rótulos legíveis: "Não"/"Sim" e, no caso de `Sexo`, "Feminino"/"Masculino". O `rename_categories` associa os rótulos na ordem dos valores (0 recebe o primeiro rótulo e 1 o segundo).

```python
# Binárias: troca 0/1 por rótulos legíveis. Sexo usa Feminino/Masculino; as demais, Não/Sim.
# rename_categories associa os rótulos na ordem dos valores (0 = 1º rótulo, 1 = 2º)
for coluna in colunas_binarias:
    if coluna != "Sexo":
        df_diabetes_tratado[coluna] = pd.Categorical(df_diabetes_tratado[coluna]).rename_categories(["Não", "Sim"])
    else:
        df_diabetes_tratado[coluna] = pd.Categorical(df_diabetes_tratado[coluna]).rename_categories(["Feminino", "Masculino"])
```

O `info()` confirma que as colunas binárias agora são do tipo `category`.

```python
df_diabetes_tratado.info()
```

## Variáveis categóricas ordinais

As colunas a seguir têm uma ordem natural. Por isso usamos `ordered=True`, para que o pandas respeite essa hierarquia em ordenações e gráficos.

`SaudeGeral`: vai de "Excelente" a "Ruim".

```python
# Variável ordinal: ordered=True preserva a hierarquia entre as categorias
df_diabetes_tratado["SaudeGeral"] = pd.Categorical(
    df_diabetes_tratado["SaudeGeral"],
    ordered=True
).rename_categories(["Excelente", "Muito Boa", "Boa", "Aceitável", "Ruim"])

df_diabetes_tratado["SaudeGeral"].head()
```

`FaixaIdade`: 13 faixas etárias, de "18 a 24 anos" até "80 ou mais anos".

```python
# Variável ordinal: ordered=True preserva a hierarquia entre as categorias
df_diabetes_tratado["FaixaIdade"] = pd.Categorical(
    df_diabetes_tratado["FaixaIdade"],
    ordered=True
).rename_categories([
    "18 a 24 anos",
    "25 a 29 anos",
    "30 a 34 anos",
    "35 a 39 anos",
    "40 a 44 anos",
    "45 a 49 anos",
    "50 a 54 anos",
    "55 a 59 anos",
    "60 a 64 anos",
    "65 a 69 anos",
    "70 a 74 anos",
    "75 a 79 anos",
    "80 ou mais anos"
])

df_diabetes_tratado["FaixaIdade"].head()
```

`Ensino`: 6 níveis de escolaridade, de "Sem estudo" até "Faculdade ou +".

```python
# Variável ordinal: ordered=True preserva a hierarquia entre as categorias
df_diabetes_tratado["Ensino"] = pd.Categorical(
    df_diabetes_tratado["Ensino"],
    ordered=True
).rename_categories([
    "Sem estudo",
    "Primário",
    "Secundário incompleto",
    "Secundário completo",
    "Faculdade incompleta ou Técnico",
    "Faculdade ou +",
    ]
)

df_diabetes_tratado["Ensino"].head()
```

`FaixaRenda`: 8 faixas de renda (em dólares), de "Menos de $10.000" até "$75.000 ou mais".

```python
# Variável ordinal: ordered=True preserva a hierarquia entre as categorias
df_diabetes_tratado["FaixaRenda"] = pd.Categorical(
    df_diabetes_tratado["FaixaRenda"],
    ordered=True
).rename_categories([
    "Menos de $10.000",
    "$10.000 a $14.999",
    "$15.000 a $19.999",
    "$20.000 a $24.999",
    "$25.000 a $34.999",
    "$35.000 a $49.999",
    "$50.000 a $74.999",
    "$75.000 ou mais"
])

df_diabetes_tratado["FaixaRenda"].head()
```

Verificando os tipos de dados depois das conversões.

```python
df_diabetes_tratado.info()
```

Conferindo as primeiras linhas da base tratada.

```python
df_diabetes_tratado.head()
```

## Variáveis numéricas

Resumo estatístico das três variáveis numéricas.

```python
df_diabetes_tratado.describe()
```

Pontos de atenção: o `IMC` tem máximo de 98, muito acima do 3º quartil (33), o que indica valores extremos (outliers). Já em `DiasProblemasMentais` e `DiasProblemasFisicos` a mediana é 0, ou seja, mais da metade das pessoas não relatou nenhum dia com problemas.

Verifica se `DiasProblemasFisicos` contém apenas números inteiros. O resultado `True` permite converter a coluna para inteiro sem perder informação.

```python
df_diabetes_tratado['DiasProblemasFisicos'].apply(float.is_integer).all()
```

Lista as colunas numéricas, que serão convertidas na próxima etapa.

```python
colunas_numericas = df_diabetes_tratado.select_dtypes(include=["number"]).columns.to_list()

colunas_numericas
```

Reduz o tipo numérico para o menor inteiro possível (`downcast="integer"`), economizando memória. O `IMC`, por exemplo, passa de `float64` para `int8`. Os valores do `describe()` anterior continuam os mesmos: só o tipo de dado muda.

```python
# downcast='integer' escolhe o menor tipo inteiro que comporta os valores (ex.: IMC vira int8)
for coluna in colunas_numericas:
    df_diabetes_tratado[coluna] = pd.to_numeric(
        df_diabetes_tratado[coluna],
        downcast="integer"
    )

df_diabetes_tratado.info()
```

## Resumo das variáveis categóricas

Quantidade de registros, valores únicos, valor mais frequente (`top`) e sua frequência (`freq`).

```python
df_diabetes_tratado.describe(exclude="number")
```

`Diabetes` tem 35.346 registros "Não" entre 70.692, portanto a base é balanceada (50% de cada classe).

## Salvando a base tratada

Exporta para `.parquet`, formato que preserva os tipos (`category`, `int8`) e que será lido pelos próximos notebooks.

```python
# Parquet preserva os tipos (category, int8); um CSV perderia essa informação
df_diabetes_tratado.to_parquet(DADOS_TRATADOS, index=False)
```
